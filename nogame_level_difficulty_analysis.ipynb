### Status Validation

To validate the meaning of `status`, I compared outcome patterns and subsequent player behavior across status values.

The analysis strongly supports `status = 1` as a successful level completion. Other status values (`0`, `2`, and `-1`) are therefore treated conservatively as non-win outcomes rather than being assigned specific meanings without documentation.

```python
con.execute(f"""
SELECT
    status,

    COUNT(*) AS rows,

    SUM(CASE WHEN unsolved_items = 0 THEN 1 ELSE 0 END) AS solved_rows,

    ROUND(
        100.0 * SUM(CASE WHEN unsolved_items = 0 THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS solved_pct,

    AVG(playtime) AS avg_playtime,
    AVG(move) AS avg_move,
    AVG(revive) AS avg_revive

FROM read_csv_auto('{data_path}/level_end.csv')

GROUP BY status
ORDER BY status
""").df()
```

```python
status_behavior = con.execute(f"""
WITH all_events AS (

    SELECT
        user_pseudo_id,
        event_timestamp,
        level_index,
        status,
        'end' AS event_type
    FROM read_csv_auto('{data_path}/level_end.csv')

    UNION ALL

    SELECT
        user_pseudo_id,
        event_timestamp,
        level_index,
        NULL AS status,
        'start' AS event_type
    FROM read_csv_auto('{data_path}/level_start.csv')
),

ordered AS (
    SELECT
        *,
        LEAD(event_type) OVER (
            PARTITION BY user_pseudo_id
            ORDER BY event_timestamp
        ) AS next_event_type,

        LEAD(level_index) OVER (
            PARTITION BY user_pseudo_id
            ORDER BY event_timestamp
        ) AS next_level_index

    FROM all_events
)

SELECT
    status,
    COUNT(*) AS rows,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN next_event_type = 'start'
                 AND next_level_index = level_index
                THEN 1 ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS immediate_retry_pct,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN next_event_type = 'start'
                 AND next_level_index = level_index + 1
                THEN 1 ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS immediate_next_level_pct

FROM ordered
WHERE event_type = 'end'

GROUP BY status
ORDER BY status
""").df()

status_behavior
```

## 4. Level Difficulty Analysis

Level difficulty is evaluated across the first 500 progression levels.

`level_index` is used to define progression scope, while `level_id`
identifies the level content being evaluated.

Multiple gameplay metrics are used to capture different dimensions of
difficulty, including first-attempt win rate, overall attempt win rate,
average attempts, playtime, and revive usage.

```python
level_metrics = con.execute("""
SELECT
    level_id,

    COUNT(*) AS attempts,

    COUNT(DISTINCT user_pseudo_id) AS users,

    ROUND(
        AVG(CASE WHEN status = 1 THEN 1.0 ELSE 0 END),
        4
    ) AS attempt_win_rate,

    ROUND(
        AVG(
            CASE
                WHEN attempt = 1
                THEN CASE WHEN status = 1 THEN 1.0 ELSE 0 END
            END
        ),
        4
    ) AS first_attempt_win_rate,

    ROUND(AVG(attempt), 2) AS avg_attempt,

    MEDIAN(attempt) AS median_attempt,

    ROUND(AVG(playtime), 2) AS avg_playtime,

    ROUND(
        AVG(CASE WHEN revive > 0 THEN 1.0 ELSE 0 END),
        4
    ) AS revive_rate,

    ROUND(AVG(unsolved_items), 2) AS avg_unsolved_items,

    ROUND(AVG(move), 2) AS avg_move

FROM '/content/level_end.parquet'

WHERE level_index BETWEEN 1 AND 500

GROUP BY level_id
ORDER BY level_id
""").df()

level_metrics.head()
```

```python
level_metrics[
    level_metrics["users"] >= 100
].sort_values(
    "first_attempt_win_rate"
).head(20)
```

```python
level_metrics[
    level_metrics["users"] >= 100
].sort_values(
    "avg_attempt",
    ascending=False
).head(20)
```

## 5. Retry Behavior

To understand whether difficult levels immediately discourage players,
I analyze player behavior after a non-win outcome.

The key metric is `retry_after_fail_rate`, which measures the share of
failed attempts followed by another attempt on the same progression level.

```python
journey_metrics = con.execute("""
WITH events AS (
    SELECT
        user_pseudo_id,
        end_ts,
        TRY_CAST(level_index AS INTEGER) AS level_index,
        TRY_CAST(level_id AS INTEGER) AS level_id,
        status
    FROM '/content/level_end.parquet'
    WHERE TRY_CAST(level_index AS INTEGER) BETWEEN 1 AND 500
),

sequenced AS (
    SELECT
        *,
        LEAD(level_index) OVER (
            PARTITION BY user_pseudo_id
            ORDER BY end_ts
        ) AS next_level_index
    FROM events
),

level_retry AS (
    SELECT
        level_id,

        COUNT(*) FILTER (
            WHERE status != 1
        ) AS failed_attempts,

        COUNT(*) FILTER (
            WHERE status != 1
              AND next_level_index = level_index
        ) AS retries_after_fail

    FROM sequenced
    GROUP BY level_id
)

SELECT
    level_id,
    failed_attempts,
    retries_after_fail,

    ROUND(
        retries_after_fail * 1.0
        / NULLIF(failed_attempts, 0),
        4
    ) AS retry_after_fail_rate

FROM level_retry
ORDER BY level_id
""").df()

journey_metrics.head()
```

```python
level_analysis = level_metrics.merge(
    journey_metrics,
    on="level_id",
    how="left"
)

level_analysis.head(20)
```

## 6. Progression Drop-off Analysis

To measure progression friction, I calculate the share of players who reached
a progression level but did not appear at the next progression level.

The analysis is limited to progression positions 1–500 to keep the scope
consistent with the difficulty analysis.

```python
progression_drop_all = con.execute("""
WITH reached AS (
    SELECT DISTINCT
        user_pseudo_id,
        TRY_CAST(level_index AS INTEGER) AS level_index
    FROM '/content/level_start.parquet'
    WHERE TRY_CAST(level_index AS INTEGER) BETWEEN 1 AND 500
),

level_reach AS (
    SELECT
        level_index,
        COUNT(DISTINCT user_pseudo_id) AS users_reached
    FROM reached
    GROUP BY level_index
),

next_reach AS (
    SELECT
        level_index - 1 AS level_index,
        COUNT(DISTINCT user_pseudo_id) AS users_reached_next
    FROM reached
    GROUP BY level_index
)

SELECT
    l.level_index,
    l.users_reached,
    COALESCE(n.users_reached_next, 0) AS users_reached_next,

    ROUND(
        1.0 - COALESCE(n.users_reached_next, 0) * 1.0 / l.users_reached,
        4
    ) AS dropoff_rate

FROM level_reach l

LEFT JOIN next_reach n
    USING(level_index)

WHERE l.users_reached >= 100

ORDER BY l.level_index
""").df()

progression_drop_all.head(10)
```

```python
progression_drop_all.sort_values(
    "dropoff_rate",
    ascending=False
).head(20)
```

### Map Progression Position to Level Content

`level_index` represents the player's progression position, while `level_id`
identifies the level content.

For each progression position, the most frequently observed `level_id`
is used as the representative content mapping.

```python
level_mapping = con.execute("""
WITH mapping AS (
    SELECT
        TRY_CAST(level_index AS INTEGER) AS level_index,
        TRY_CAST(level_id AS INTEGER) AS level_id,
        COUNT(*) AS n,

        ROW_NUMBER() OVER (
            PARTITION BY TRY_CAST(level_index AS INTEGER)
            ORDER BY COUNT(*) DESC
        ) AS rn

    FROM '/content/level_start.parquet'

    WHERE TRY_CAST(level_index AS INTEGER) BETWEEN 1 AND 500

    GROUP BY
        TRY_CAST(level_index AS INTEGER),
        TRY_CAST(level_id AS INTEGER)
)

SELECT
    level_index,
    level_id

FROM mapping

WHERE rn = 1

ORDER BY level_index
""").df()

level_mapping.head(10)
```

### Priority Scoring

To support level prioritization, I created a heuristic score combining:

- 40% difficulty
- 40% progression drop-off
- 20% player exposure

This score is used only as a prioritization aid, not as a causal or predictive model.

```python
drop_with_content = progression_drop_all.merge(
    level_mapping,
    on="level_index",
    how="left"
)

drop_with_content.head()
```

```python
priority_levels = drop_with_content.merge(
    level_analysis,
    on="level_id",
    how="left"
)

priority_levels[
    [
        "level_id",
        "users_reached",
        "dropoff_rate",
        "first_attempt_win_rate",
        "avg_attempt",
        "revive_rate",
        "retry_after_fail_rate"
    ]
].sort_values(
    ["dropoff_rate", "first_attempt_win_rate"],
    ascending=[False, True]
).head(20)
```

```python
rank_df = priority_levels.copy()

rank_df["difficulty_score"] = 1 - rank_df["first_attempt_win_rate"]
rank_df["exposure_score"] = rank_df["users_reached"] / rank_df["users_reached"].max()

rank_df["priority_score"] = (
    0.4 * rank_df["difficulty_score"] +
    0.4 * rank_df["dropoff_rate"] / rank_df["dropoff_rate"].max() +
    0.2 * rank_df["exposure_score"]
)

rank_df[
    [
        "level_id",
        "users_reached",
        "first_attempt_win_rate",
        "avg_attempt",
        "dropoff_rate",
        "priority_score"
    ]
].sort_values(
    "priority_score",
    ascending=False
).head(10)
```
